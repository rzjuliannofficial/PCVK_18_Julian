# Langkah (Step) 2:

Baca Image dari URLs - Read Image From URLs

--Bahasa--
Pada Langkah ini kita akan membaca image dari URLs, menampilkannya menggunakan OpenCV, perhatikan ada perbedaan ketika membuk image dalam RGB dan BGR. Channel Warna default untuk OpenCV adalah BGR.

--English--
In this step we will read images from urls, and display them using openCV, please note the difference when reading image in RGB and BGR format. The default input color channels are in BGR format for openCV. RGB?

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placecats.com/800/571"]

# Baca dan tampilkan image
# loopp pada tiap url image, bebeerapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                           #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)  #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   #convert image to RGB
  final_frame = cv.hconcat((image, image_2))       #concat image
  cv2_imshow(final_frame)                          #display image
  print('\n')
```

# Step 3

Melihat ukuran file Image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

# Langkah (Step) 4
mengakses pixel dengan memberi garis diagonal menyilang

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

#TUGAS PRAKTIKUM D2

1. Buat garis vertikal dan garis menyilang diagonal pada image keluaran
2. Buat garis horisontal berwarna putih dibagian tengah gambar dengan
panjang tertentu
3. Buat kotak menggunakan kumpulan pixel warna putih di sembarang tempat dalam gambar

```python
#1 Garis Vertikal
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_5 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range(tinggi) :
  image_5[x,int((lebar)/2)] = [255,255,255]

final_frame_1 = cv.hconcat((image_2, image_5))
cv2_imshow(final_frame_1)

# Garis Diagonal
image_6 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
for x in range(tinggi) :
  image_6[x, int(x * (lebar / tinggi))] = [255, 255, 255]

  # Diagonal 2 (kanan atas ke kiri bawah)
  image_6[x, int((tinggi - 1 - x) * (lebar / tinggi))] = [255, 255, 255]

final_frame_2 = cv.hconcat((image_2, image_6))
cv2_imshow(final_frame_2)

#2 Garis ditengah dengan panjang tertentu
image_7 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
y_awal = int((lebar)/2 - 100)
y_akhir = int((lebar)/2 + 50)
for y in range (y_awal, y_akhir):
  image_7[int((tinggi)/2),y] = [255,255,255]

final_frame_3 = cv.hconcat((image_2, image_7))
cv2_imshow(final_frame_3)
```

```python
#3 Kumpulan pixel disembarang tempat
image_kotak = image_2.copy()

for b in range(80, 160):       # Loop baris (tinggi)
  for k in range(120, 200):    # Loop kolom (lebar)
    image_kotak[b, k] = [255, 255, 255]

final_frame_4 = cv.hconcat((image_2, image_kotak))
cv2_imshow(final_frame_4)
```
